####    갱신 뉴스 기사에 대해 tokenizing 진행

```python
import re
import os
import multiprocessing
import pandas as pd
from pathlib import Path
from joblib import Parallel, delayed

#   경로 내 파일 확인
path_001 = r'./data'

"""해당 파일에서는 기존 파일에서 누락된 날짜도 추가"""
"""기존 파일(parquet) 날짜 범위 확인"""
check_list_001 = list(Path(path_001).rglob('df_tokenized*.parquet'))

"""경로 내 파일 불러오기"""
#    경로 내 parquet파일 일괄 불러오기 함수 생성
def read_parquet_01(file_01):
    return pd.read_parquet(file_01)

#   병렬 처리 설정
cl = multiprocessing.cpu_count() - 1

#   해당 경로 내 파일을 list 형태로 불러오기
check_ls_001 = Parallel(n_jobs = cl)(delayed(read_parquet_01)(file_01) for file_01 in check_list_001)
```

```python
#   list 형태의 파일 내용을 행 기반 데이터 프레임 형태로 결합
check_001 = pd.concat(check_ls_001, axis = 0)

#   Target_Date(날짜) 열 데이터 속성 확인
print(type(check_001['Target_Date']))

#   기존 parquet 파일 날짜 범위 확인
print(check_001['Target_Date'].describe())
```

```python
#   네이버 경제뉴스 기사 보관 파일 경로 지정
path_002 = Path(r'D:/네이버뉴스/기사/경제')

#   경로 내 폴더 경로 추출
#   iterdir(): 경로 내 모든 요소
#   is_dir() : 폴더 여부 확인
df_list_001 = [p for p in path_002.iterdir() if p.is_dir()]

#   경로 확인
for d in df_list_001:
    print(d)

#   필요 경로 추출
df_list_002 = df_list_001[4:6]
print(df_list_002)
```

```python
#   필요 경로 내 파일 목록 추출
df_list_003 = [file_02 for p in df_list_002 for file_02 in p.rglob('네이버_경제*.csv')]

for check in df_list_003:
    print(check)

#   필요 날짜 범위에 해당하는 파일 추출
target_date = '260522'
df_list_004 = []

for p in df_list_003:
    #   파일 명 내 일자 추출
    match = re.search(r'\d{6}', p.name)
    #   일자 비교
    if match.group() >= target_date:
        df_list_004.append(p)
    
df_list_004
```

```python
#    경로 내 csv파일 일괄 불러오기 함수 생성
def read_csv_01(f_02):
    return pd.read_csv(f_02)

df_ls_001 = Parallel(n_jobs = cl)(delayed(read_csv_01)(file_01) for file_01 in df_list_004)

df_001 = pd.concat(df_ls_001, axis = 0, ignore_index = True)
df_001['일자'] = pd.to_datetime(df_001['일자']).astype('datetime64[ns]')

df_001['일자'].describe()
```

```python
#   결측값 확인
for col_01 in df_001.columns.tolist():
    print(df_001.isnull()[col_01].value_counts())

#   '본문'열 결측 제거
df_001.dropna(subset = ['본문'], inplace = True)

#   결측처리 확인
df_001['본문'].isnull().value_counts(dropna=False)
```

```python
#   중복 확인
for col_02 in df_001.columns.tolist():
    print(df_001[col_02].duplicated().value_counts())

#   '제목' 열 기준 중복값 처리
df_001.drop_duplicates(subset = ['제목'], inplace = True)

#   중복처리 확인
df_001['제목'].duplicated().value_counts()
```

```python
df_002 = df_001.copy()

#   '수정'열 생성: 특수문자 처리
df_002['수정'] = df_001['본문'].apply(lambda x: re.sub('[^ A-Za-z0-9ㄱ-ㅣ가-힣]', ' ', x))

#   '수정'열: 문둔 공백에 대해 ""으로 치환
df_002['수정'] = df_002['수정'].apply(lambda x: re.sub('^ +', '', x))

#   '수정'열: 공란을 결측치로 변환
df_002['수정'] = df_002['수정'].replace('', None)
```

```python
#   '수정'열: 결측 확인
print(df_002['수정'].isnull().value_counts(dropna = False))

#   '수정'열: 중복 확인
print(df_002['수정'].duplicated().value_counts(dropna = False))

#   '수정'열: 중복 제거
df_002.drop_duplicates(subset = ['수정'], inplace = True)

#   '수정'열: 중복 제거 확인
print(df_002['수정'].duplicated().value_counts(dropna = False))
```

```python
#   kiwipiepy 활용, 토큰화 진행
from kiwipiepy import Kiwi

kiwi = Kiwi()

stnd_pos = ['NNG', 'NNP', 'VV', 'VA']
stop_words = []

def kiwi_tokenizer(article_01):
    text_01 = str(article_01)
    tokens_01 = kiwi.tokenize(text_01)
    token_list_01 = [token.form for token in tokens_01
                    # 품사 설정
                    if token.tag in stnd_pos and
                    # 불용어 제외
                    token.form not in stop_words
                    ]
    return token_list_01

from tqdm import tqdm
tqdm.pandas()
```

```python
df_002['tokens'] = df_002['수정'].progress_apply(kiwi_tokenizer)
```

```python
df_002['일자'] = pd.to_datetime(df_002['일자']).astype('datetime64[ns]')

min_date = df_002['일자'].min().strftime('%y%m%d')
max_date = df_002['일자'].max().strftime('%y%m%d')

df_002.to_parquet(f'./data/df_tokenized_{min_date}_{max_date}.parquet')
```
